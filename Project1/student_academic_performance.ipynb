Detect outliers

```python
Q1 = df["Math Score"].quantile(0.25)
Q3 = df["Math Score"].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = df[(df["Math Score"] < lower) | (df["Math Score"] > upper)]

outliers
```

Treat outliers

```python
df_no_outliers = df[(df["Math Score"] >= lower) &
                    (df["Math Score"] <= upper)]

print("Original Shape :", df.shape)
print("After Removing Outliers :", df_no_outliers.shape)
```

Bar chart of average math score by gender

```python
avg_score = df.groupby("Gender")["Math Score"].mean()

plt.figure(figsize=(6,4))
plt.bar(avg_score.index, avg_score.values)

plt.title("Average Math Score by Gender")
plt.xlabel("Gender")
plt.ylabel("Average Math Score")

plt.show()
```
