Chect missing values

```python
df.isnull().sum()
```

```python
df['fixed acidity'] = df['fixed acidity'].fillna(df['fixed acidity'].mode()[0])
```

```python
df['citric acid'] = df['citric acid'].fillna(df['citric acid'].mode()[0])
```

```python
df['residual sugar'] = df['residual sugar'].fillna(df['residual sugar'].mode()[0])
```

```python
df['volatile acidity'] = df['volatile acidity'].fillna(df['volatile acidity'].mode()[0])
```

```python
df['chlorides'] = df['chlorides'].fillna(df['chlorides'].mode()[0])
```

```python
df['pH'] = df['pH'].fillna(df['pH'].mode()[0])
```

```python
df['sulphates'] = df['sulphates'].fillna(df['sulphates'].mode()[0])
```

```python
df.isnull().sum()
```

Chect duplicates values

```python
df.duplicated().sum()
```

```python
df = df.drop_duplicates()
```

Calculate correlation matrix

```python
correlation = df.corr(numeric_only=True)

correlation
```

Display Heatmap

```python
plt.figure(figsize=(12,8))

sns.heatmap(correlation,
            annot=True,
            cmap='coolwarm')

plt.title("Correlation Matrix")
plt.show()
```

Identify Highly Correlated Features

```python
high_corr = correlation["quality"].sort_values(ascending=False)

high_corr
```

```python
high_corr[abs(high_corr) > 0.3]
```

Create Visualizaton

Histogram

```python
plt.figure(figsize=(6,4))

plt.hist(df["quality"], bins=10)

plt.title("Wine Quality Distribution")

plt.xlabel("Quality")

plt.ylabel("Count")

plt.show()


```

Bar chart

```python
df["quality"].value_counts().sort_index().plot(kind="bar")

plt.title("Quality Counts")

plt.xlabel("Quality")

plt.ylabel("Count")

plt.show()
```

Scatter plot

```python
plt.figure(figsize=(6,4))

plt.scatter(df["alcohol"],
            df["quality"])

plt.xlabel("Alcohol")

plt.ylabel("Quality")

plt.title("Alcohol vs Quality")

plt.show()
```

Line chart

```python
plt.figure(figsize=(8,4))

plt.plot(df["alcohol"])

plt.title("Alcohol Values")

plt.xlabel("Index")

plt.ylabel("Alcohol")

plt.show()
```
